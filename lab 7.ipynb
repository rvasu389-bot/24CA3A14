## In-Lab Program

The following program demonstrates the experiment described in the workbook. Run the cell after installing any required packages or downloading required models.

```python
import gensim.downloader as api
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

# Downloads the pre-trained model on first run; internet is required.
model = api.load("glove-wiki-gigaword-50")

s1 = "I love machine learning"
s2 = "I enjoy artificial intelligence"

def sentence_vector(sentence):
    words = sentence.lower().split()
    vectors = [model[w] for w in words if w in model]
    return np.mean(vectors, axis=0)

v1 = sentence_vector(s1)
v2 = sentence_vector(s2)
score = cosine_similarity([v1], [v2])[0][0]

print("Sentence 1:", s1)
print("Sentence 2:", s2)
print("Cosine Similarity:", round(float(score), 2))
```

## Post-Lab Answer

GloVe stands for Global Vectors for Word Representation. Its vectors are learned using global word co-occurrence statistics from a corpus. Word2Vec commonly learns embeddings by predicting target or context words, while GloVe learns from aggregated co-occurrence information. A pre-trained GloVe model has already been trained on a large dataset and can be loaded for downstream NLP tasks. Averaging word vectors is a simple way to obtain sentence vectors, though it does not fully represent word order.
