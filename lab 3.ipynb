## In-Lab Program

The following program demonstrates the experiment described in the workbook. Run the cell after installing any required packages or downloading required models.

```python
from sklearn.feature_extraction.text import CountVectorizer
from wordcloud import WordCloud
import matplotlib.pyplot as plt

corpus = ["I love NLP", "I love Python", "NLP is fun"]
words = sorted(set(" ".join(corpus).lower().split()))
print("Vocabulary:", words)

for word in words:
    print(word, [int(w == word) for w in words])

vectorizer = CountVectorizer()
matrix = vectorizer.fit_transform(corpus)
print("Features:", vectorizer.get_feature_names_out())
print("BoW Matrix:\n", matrix.toarray())

combined_text = " ".join(corpus)
cloud = WordCloud(width=600, height=300, background_color="white")
cloud.generate(combined_text)
plt.figure(figsize=(8, 4))
plt.imshow(cloud, interpolation="bilinear")
plt.axis("off")
plt.show()
```

## Post-Lab Answer

One-Hot Encoding creates a separate binary vector for each unique word. It indicates word identity but does not show semantic similarity. Bag of Words (BoW) represents each document using counts of vocabulary words and therefore captures frequency. In this corpus, `i`, `love`, and `nlp` each occur twice. The Word Cloud provides a visual version of frequency information, while the BoW matrix gives exact numerical values. Higher-frequency words generally appear larger in the Word Cloud. Both representations are useful for basic text analysis, but they do not inherently capture word meaning.
