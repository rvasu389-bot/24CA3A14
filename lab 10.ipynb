**5. What is NER?**

Named Entity Recognition identifies and labels entities such as people, organizations, locations, dates, and money in text.

**6. What is a pre-trained BERT model?**

BERT is a transformer-based language model trained on large text collections to learn contextual representations. A pre-trained model has learned general language patterns before being adapted to a task.

**7. Why is fine-tuning required for BERT?**

Fine-tuning trains the pre-trained model on labeled task-specific examples so it can adapt its learned representations to a particular task such as NER.

## In-Lab Program

The following program demonstrates the experiment described in the workbook. Run the cell after installing any required packages or downloading required models.

```python
from transformers import pipeline

# Uses a pre-trained BERT-based NER model for inference.
# This demonstrates entity prediction; it does not fine-tune the model.
ner = pipeline(
    "ner",
    model="dslim/bert-base-NER",
    aggregation_strategy="simple"
)

sentences = [
    "Sundar Pichai is the CEO of Google.",
    "Microsoft was founded by Bill Gates.",
    "The conference will be held in Chennai on 20 July 2026.",
    "Elon Musk visited India last week."
]

for sentence in sentences:
    print("\nSentence:", sentence)
    for entity in ner(sentence):
        print(entity["word"], ":", entity["entity_group"])
```
