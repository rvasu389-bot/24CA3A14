# Experiment 9: Build an LSTM Model for Sentiment Classification

**Aim:** To build and evaluate an LSTM model for classifying text sentiment as Positive or Negative.

**Course:** Natural Language Processing and Large Language Models (24CA31A4)

## Pre-Lab Answers

**1. Assign sentiment labels to the six reviews.**

Positive: `The product quality is excellent`; `I am very happy with the service`; `Excellent customer support`. Negative: `The delivery was very slow`; `The product is damaged`; `Very disappointed with the purchase`.

**2. How is an LSTM model trained?**

Reviews are tokenized, converted into integer sequences, padded to equal lengths, and passed to an embedding layer and an LSTM network. The model learns from labeled positive and negative examples.

**3. Predict the sentiment of 'The product is very good and easy to use.'**

Expected sentiment: Positive, because the sentence expresses a favorable opinion.

## In-Lab Program

The following program demonstrates the experiment described in the workbook. Run the cell after installing any required packages or downloading required models.

```python
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense

reviews = [
    "product quality is excellent",
    "happy with the service",
    "delivery was very slow",
    "product is damaged",
    "excellent customer support",
    "disappointed with purchase"
]
labels = [1, 1, 0, 0, 1, 0]  # 1 = Positive, 0 = Negative

tokenizer = Tokenizer(num_words=1000, oov_token="<OOV>")
tokenizer.fit_on_texts(reviews)
X = tokenizer.texts_to_sequences(reviews)
X = pad_sequences(X, maxlen=8, padding="post")

model = Sequential([
    Embedding(1000, 16, input_length=8),
    LSTM(16),
    Dense(1, activation="sigmoid")
])
model.compile(optimizer="adam", loss="binary_crossentropy",
              metrics=["accuracy"])
model.fit(X, labels, epochs=20, verbose=0)

test = ["the product is very good and easy to use"]
test_seq = tokenizer.texts_to_sequences(test)
test_pad = pad_sequences(test_seq, maxlen=8, padding="post")
prediction = model.predict(test_pad, verbose=0)[0][0]
print("Sentiment:", "Positive" if prediction >= 0.5 else "Negative")
```

## Post-Lab Answer

The Embedding layer maps integer word IDs to dense numerical vectors that can be learned during training. The LSTM layer processes the sequence and learns patterns that help classify sentiment. LSTM networks use gates and a cell state to retain useful information over longer sequences. Compared with a traditional RNN, they can better learn long-term dependencies and help reduce the vanishing-gradient problem. A small training set is useful for demonstrating the architecture but is not sufficient for a reliable production model.
