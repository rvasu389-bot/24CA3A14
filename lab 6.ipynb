# Experiment 6: Word Embeddings Using Word2Vec

**Aim:** To generate word embeddings using Word2Vec and measure semantic similarity between sentences.

## Pre-Lab Answers

**1. Which words have similar meanings?**

`Love` and `enjoy` have related meanings in the sample sentences. `Cats` and `dogs` are related because both are animals and occur in similar contexts.

**2. How can Word2Vec represent words numerically?**

Word2Vec learns a dense vector for each word based on surrounding words in a training corpus. Words used in similar contexts may have similar vectors.

**3. Which architecture predicts the missing word?**

CBOW predicts a target word from its surrounding context words.

**4. Which architecture predicts surrounding words?**

Skip-Gram uses a target word to predict nearby context words.

## In-Lab Program

The following program demonstrates the experiment described in the workbook. Run the cell after installing any required packages or downloading required models.

```python
from gensim.models import Word2Vec
from sklearn.metrics.pairwise import cosine_similarity
import numpy as np

sentences = [
    ["i", "love", "machine", "learning"],
    ["i", "enjoy", "artificial", "intelligence"],
    ["cats", "drink", "milk"],
    ["dogs", "drink", "milk"]
]
model = Word2Vec(sentences, vector_size=50, window=3,
                 min_count=1, workers=1, seed=42)

s1 = ["i", "love", "machine", "learning"]
s2 = ["i", "enjoy", "artificial", "intelligence"]
v1 = np.mean([model.wv[w] for w in s1], axis=0)
v2 = np.mean([model.wv[w] for w in s2], axis=0)
score = cosine_similarity([v1], [v2])[0][0]

print("Sentence 1:", " ".join(s1))
print("Sentence 2:", " ".join(s2))
print("Cosine Similarity:", round(float(score), 2))
```

## Post-Lab Answer

Word2Vec is a neural-network-based technique that learns dense word embeddings from context. Compared with Bag of Words, embeddings can represent relationships between words instead of treating each word as an unrelated feature. CBOW predicts a target word from surrounding context, while Skip-Gram predicts surrounding context words from a target word. Word vectors can be compared using cosine similarity. With a very small corpus, the learned vectors are only illustrative and should not be treated as reliable general-language representations.
