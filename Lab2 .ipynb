```python
import nltk
import string
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
text = "Dr. Ravi is studying machine Learning at Anna University in Chennai."
```

```python
tokens = word_tokenize(text)
```

```python
stop_words = set (stopwords.words('english'))
filtered_tokens = [
    word
    for word in tokens
    if word.lower() not in stop_words 
    and word not in string.punctuation]
```

```python
print("Original Tokens:")
print(tokens)

print("\nTokens after stop-word & Punctuation Remoal:")
print(filtered_tokens)
```

```python
Step 3: Stemming
```

```python
import nltk
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer
import string
```

```python
nltk.download('punkt')
nltk.download('stopwords')
```

```python
text = "Dr. Ravi is studying Machine Learning at Anna University in Chennai."
```

```python
tokens = word_tokenize(text)
```

```python
stop_words = set(stopwords.words('english'))

filtered_tokens = [
    word.replace(".", "")
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]
```

```python
stemmer = PorterStemmer()

stemmed_tokens = [
    stemmer.stem(word.lower())
    for word in filtered_tokens
]
```

```python
print("Tokens after Stop-word & Punctuation Removal:")
print(filtered_tokens)

print("\nStemmed Tokens:")
print(stemmed_tokens)
```

```python
4. Lemmatization
```

```python
import nltk
import string

from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
```

```python
nltk.download('punkt')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
text = "Dr. Ravi is studying Machine Learning at Anna University in Chennai."
```

```python
tokens = word_tokenize(text)
```

```python
stop_words = set(stopwords.words("english"))

filtered_tokens = [
    word.replace(".", "")
    for word in tokens
    if word.lower() not in stop_words
    and word not in string.punctuation
]
```

```python
lemmatizer = WordNetLemmatizer()

lemmatized_tokens = [
    lemmatizer.lemmatize(word)
    for word in filtered_tokens
]
```

```python
print("Filtered Tokens:")
print(filtered_tokens)

print("\nLemmatized Tokens:")
print(lemmatized_tokens)
```
