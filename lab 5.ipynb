## Pre-Lab Answers

**1. What is text classification?**

Text classification is an NLP task that assigns a text item to one or more predefined classes. Examples include spam detection, sentiment analysis, and news categorization.

**2. What are the two classes in spam detection?**

The classes are Spam and Not Spam.

**3. Which words may indicate that an email is spam?**

Words such as `free`, `prize`, `lottery`, `claim`, and `won` may indicate spam, although context is important.

**4. Which algorithm is commonly used for spam detection?**

Naïve Bayes is a commonly used algorithm for spam detection because it estimates class probabilities from word features.

**5. What is sentiment analysis?**

Sentiment analysis determines whether a text expresses a positive, negative, or neutral opinion.

**6. What are the sentiment classes?**

Positive, Negative, and Neutral.

**7. Which algorithm is used for sentiment analysis in this experiment?**

Support Vector Machine (SVM), which learns a decision boundary between classes.

**8. Which review is positive?**

`This phone is excellent.`

**9. Which review is negative?**

## In-Lab Program

The following program demonstrates the experiment described in the workbook. Run the cell after installing any required packages or downloading required models.

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import MultinomialNB
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score, confusion_matrix

# Activity 1: Spam detection
emails = [
    "Win a free lottery prize", "Claim your free prize now",
    "Meeting at 10 AM tomorrow", "Project report submission is due",
    "Congratulations you won a lottery", "Let us discuss the project"
]
labels = ["Spam", "Spam", "Not Spam", "Not Spam", "Spam", "Not Spam"]

x_train, x_test, y_train, y_test = train_test_split(
    emails, labels, test_size=0.33, random_state=42, stratify=labels
)
v = TfidfVectorizer()
X_train = v.fit_transform(x_train)
X_test = v.transform(x_test)

nb = MultinomialNB().fit(X_train, y_train)
pred = nb.predict(X_test)
print("Naive Bayes Accuracy:", accuracy_score(y_test, pred))
print("Confusion Matrix:\n", confusion_matrix(y_test, pred))
print("New email prediction:",
      nb.predict(v.transform(["Win a free prize"]))[0])

# Activity 2: Sentiment classification using SVM
reviews = [
    "This phone is excellent", "Battery performance is poor",
    "The product is okay", "Amazing quality and service",
    "Very bad experience", "The item is average"
]
sentiments = ["Positive", "Negative", "Neutral",
              "Positive", "Negative", "Neutral"]

sv = TfidfVectorizer()
X = sv.fit_transform(reviews)
svm = LinearSVC().fit(X, sentiments)
print("New review sentiment:",
      svm.predict(sv.transform(["Excellent product"]))[0])
```
