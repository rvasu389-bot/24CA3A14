# Experiment 4: TF-IDF (Term Frequency–Inverse Document Frequency)

**Aim:** To convert text into weighted numerical features using TF-IDF and identify important words in documents.

**Course:** Natural Language Processing and Large Language Models (24CA31A4)

## Pre-Lab Answers

**1. List the unique words in the documents.**

For `Apple is sweet`, `Apple is red`, and `Mango is sweet`, the vocabulary is `apple, is, sweet, red, mango`.

**2. What is the vocabulary size?**

There are five unique words.

**3. Calculate Term Frequency (TF) and write the formula.**

Term Frequency measures how often a term occurs in a document relative to the document length. Formula: `TF(t,d) = count of term t in document d / total terms in document d`.

**4. Write the formula and calculate Document Frequency (DF).**

Document Frequency is the number of documents containing a term. Here, `apple` appears in D1 and D2, so DF = 2; `is` appears in all three, so DF = 3; `sweet` appears in D1 and D3, so DF = 2; `red` appears only in D2, so DF = 1; and `mango` appears only in D3, so DF = 1.

**5. Write the formula for IDF.**

A common formula is `IDF(t) = log(N / DF(t))`, where N is the total number of documents. Terms found in every document receive a lower IDF than terms found in only one document.

**6. Write the TF-IDF formula.**

`TF-IDF(t,d) = TF(t,d) × IDF(t)`. It gives higher weight to terms that are frequent in a document but not common across the entire collection.

## In-Lab Program

The following program demonstrates the experiment described in the workbook. Run the cell after installing any required packages or downloading required models.

```python
import numpy as np
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer

docs = ["Apple is sweet", "Apple is red", "Mango is sweet"]
cv = CountVectorizer()
counts = cv.fit_transform(docs).toarray()
terms = cv.get_feature_names_out()

tf = counts / counts.sum(axis=1, keepdims=True)
df = (counts > 0).sum(axis=0)
idf = np.log(len(docs) / df)

print("Vocabulary:", terms)
print("TF Matrix:\n", np.round(tf, 3))
print("DF:", dict(zip(terms, df)))
print("IDF:", dict(zip(terms, np.round(idf, 3))))

tv = TfidfVectorizer()
tfidf = tv.fit_transform(docs).toarray()
print("TF-IDF Features:", tv.get_feature_names_out())
print("TF-IDF Matrix:\n", np.round(tfidf, 3))
```
